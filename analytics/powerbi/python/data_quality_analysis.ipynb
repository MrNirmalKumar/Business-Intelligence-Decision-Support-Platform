```python
import pandas as pd
import os

file_path = os.path.join("..", "data", "Global_Superstore2.csv")

df = pd.read_csv(
    file_path,
    encoding="latin1"
)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
# Dataset Information

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Correct Date Parsing

order_dates = pd.to_datetime(
    df["Order Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

ship_dates = pd.to_datetime(
    df["Ship Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Invalid Order Dates:", order_dates.isna().sum())
print("Invalid Ship Dates:", ship_dates.isna().sum())

print("\nOrder Date Range:")
print(order_dates.min(), "to", order_dates.max())

print("\nShip Date Range:")
print(ship_dates.min(), "to", ship_dates.max())
```

```python
# Numeric Columns Validation

numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Shipping Cost"
]

for column in numeric_columns:
    print(f"\n--- {column} ---")
    print("Data Type:", df[column].dtype)
    print("Minimum Value:", df[column].min())
    print("Maximum Value:", df[column].max())
    print("Missing Values:", df[column].isna().sum())
    print("Negative Values:", (df[column] < 0).sum())
```

```python
# Categorical Values Check

categorical_columns = [
    "Ship Mode",
    "Segment",
    "Category",
    "Market",
    "Region",
    "Order Priority"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print("Unique Values:", df[column].nunique())
    print(df[column].unique())
```

```python
# Duplicate Records Check

duplicate_rows = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_rows)

if duplicate_rows == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records are present.")
```

```python
# Create a copy for data cleaning

df_clean = df.copy()

print("Cleaned dataset copy created.")
print("Shape:", df_clean.shape)
```

```python
# Convert date columns to datetime

df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"],
    format="mixed",
    dayfirst=True
)

df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"],
    format="mixed",
    dayfirst=True
)

print("Date columns converted successfully.")

print("\nOrder Date:", df_clean["Order Date"].dtype)
print("Ship Date:", df_clean["Ship Date"].dtype)
```

```python
# Convert Postal Code to nullable integer

df_clean["Postal Code"] = df_clean["Postal Code"].astype("Int64")

print("Postal Code datatype:", df_clean["Postal Code"].dtype)
print("Missing Postal Codes:", df_clean["Postal Code"].isna().sum())
```

```python
# Final Data Quality Check

print("Dataset Shape:", df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nDuplicate Rows:", df_clean.duplicated().sum())
```

```python
# Save cleaned dataset

output_path = "../data/Global_Superstore2_Cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_path)
```

```python
# Verify cleaned CSV

cleaned_df = pd.read_csv(
    "../data/Global_Superstore2_Cleaned.csv",
    encoding="latin1"
)

print("Cleaned CSV loaded successfully!")
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])

cleaned_df.head()
```

```python
# Overall Business Summary

total_sales = cleaned_df["Sales"].sum()
total_profit = cleaned_df["Profit"].sum()
total_quantity = cleaned_df["Quantity"].sum()
total_orders = cleaned_df["Order ID"].nunique()
total_customers = cleaned_df["Customer ID"].nunique()

print("===== BUSINESS SUMMARY =====")
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
```

```python
# Category-wise Sales and Profit Analysis

category_analysis = (
    cleaned_df
    .groupby("Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
    .sort_values("Sales", ascending=False)
)

category_analysis
```

```python
import matplotlib.pyplot as plt

category_analysis["Sales"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Sub-Category-wise Sales and Profit Analysis

subcategory_analysis = (
    cleaned_df
    .groupby("Sub-Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
    .sort_values("Sales", ascending=False)
)

subcategory_analysis
```

```python
# Sub-Category Sales Chart

subcategory_analysis["Sales"].sort_values().plot(
    kind="barh",
    figsize=(10, 7)
)

plt.title("Sales by Sub-Category")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()
```

```python
# Profit by Sub-Category

profit_subcategory = subcategory_analysis["Profit"].sort_values()

profit_subcategory
```

```python
# Profit by Sub-Category Chart

profit_subcategory.plot(
    kind="barh",
    figsize=(10, 7)
)

plt.title("Profit by Sub-Category")
plt.xlabel("Profit")
plt.ylabel("Sub-Category")
plt.axvline(0)
plt.tight_layout()
plt.show()
```

```python
# Year-wise Sales and Profit Analysis

df_clean["Order Year"] = df_clean["Order Date"].dt.year

yearly_analysis = (
    df_clean
    .groupby("Order Year")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
)

yearly_analysis
```

```python
# Year-wise Sales Trend

yearly_analysis["Sales"].plot(
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Year-wise Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.xticks(yearly_analysis.index)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Year-wise Profit Trend

yearly_analysis["Profit"].plot(
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Year-wise Profit Trend")
plt.xlabel("Year")
plt.ylabel("Profit")
plt.xticks(yearly_analysis.index)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CONSOLIDATED EDA ANALYSIS
# ============================================================

# 1. Segment-wise Analysis
segment_analysis = (
    df_clean
    .groupby("Segment")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique")
    )
    .sort_values("Sales", ascending=False)
)

print("===== SEGMENT ANALYSIS =====")
display(segment_analysis.round(2))


# 2. Region-wise Analysis
region_analysis = (
    df_clean
    .groupby("Region")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Sales", ascending=False)
)

print("\n===== REGION ANALYSIS =====")
display(region_analysis.round(2))


# 3. Market-wise Analysis
market_analysis = (
    df_clean
    .groupby("Market")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Sales", ascending=False)
)

print("\n===== MARKET ANALYSIS =====")
display(market_analysis.round(2))


# 4. Shipping Mode Analysis
shipping_analysis = (
    df_clean
    .groupby("Ship Mode")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Orders", ascending=False)
)

print("\n===== SHIPPING MODE ANALYSIS =====")
display(shipping_analysis.round(2))


# 5. Top 10 Customers by Sales
top_customers = (
    df_clean
    .groupby(["Customer ID", "Customer Name"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Sales", ascending=False)
    .head(10)
)

print("\n===== TOP 10 CUSTOMERS BY SALES =====")
display(top_customers.round(2))


# 6. Top 10 Products by Sales
top_products = (
    df_clean
    .groupby(["Product ID", "Product Name"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
    .sort_values("Sales", ascending=False)
    .head(10)
)

print("\n===== TOP 10 PRODUCTS BY SALES =====")
display(top_products.round(2))


# 7. Loss-making Sub-Categories
loss_subcategories = (
    subcategory_analysis[subcategory_analysis["Profit"] < 0]
    .sort_values("Profit")
)

print("\n===== LOSS-MAKING SUB-CATEGORIES =====")
display(loss_subcategories.round(2))


# 8. Discount vs Profit Summary
discount_analysis = (
    df_clean
    .groupby("Discount")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
    .sort_index()
)

print("\n===== DISCOUNT vs PROFIT =====")
display(discount_analysis.round(2))
```
